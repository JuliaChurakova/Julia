# Глава 10 Форматированный вывод

Функции **@printf** и **@sprintf** из пакета Printf.jl в языке программирования Julia используются аналогично функциям **printf** и **sprintf** из языка Си. Она принимает строку формата, а затем значения, которые нужно вставить в этот формат, и возвращает отформатированную строку.

Для получения дополнительной информации о пакете см. [документацию.](https://docs.julialang.org/en/v1/stdlib/Printf/)

```julia
using Printf
x = 3.14159
@printf("%.2f", x)
```

```julia
using Printf
h=-123.678549
k=1234567889
@printf("h=%6.4f\n" ,h)
@printf("k=%15d\n" ,k)
@printf("h=%16.9f\tk=%d\n" , h, k)
```

```julia
n = 4
m = 7
A = rand(n, m)

println("Матрица A\n")
for i in 1:n
    for j in 1:m
        @printf("%6.2f", A[i, j])
    end
    println()
end
```
